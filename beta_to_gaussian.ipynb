```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma
```

```python

def plot_gaussian_beta(d_list):

    sigma = 1  # Standard Gaussian

    x_gauss = np.linspace(-5, 5, 1000)
    gaussian = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(
        -x_gauss**2 / (2 * sigma**2)
    )

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

    for d in d_list:

        x_beta = np.linspace(-np.sqrt(d) + 0.0001,
                             np.sqrt(d) - 0.0001, 1000)

        if np.sqrt(d) > 5:
            x_beta = np.linspace(-5, 5, 1000)

        beta_pdf = (
            gamma(d / 2) /
            (np.sqrt(d) * np.sqrt(np.pi) * gamma((d - 1) / 2))
        ) * (1 - x_beta**2 / d)**((d - 3) / 2)

        # Gaussian evaluated at the same x points
        gaussian_at_x = (1 / np.sqrt(2 * np.pi)) * np.exp(-x_beta**2 / 2)

        # Plot distributions
        ax1.plot(x_beta, beta_pdf, linewidth=2, label=f'd = {d}')

        # Plot absolute error
        error = np.abs(beta_pdf - gaussian_at_x)
        ax2.plot(x_beta, error, linewidth=2, label=f'd = {d}')

    # Gaussian reference
    ax1.plot(x_gauss, gaussian, 'k--', linewidth=2, label='Standard Gaussian')

    ax1.set_xlabel('x')
    ax1.set_ylabel('Probability density')
    ax1.set_title('Beta vs. Standard Gaussian')
    ax1.grid(alpha=0.3)
    ax1.legend()

    ax2.set_xlabel('x')
    ax2.set_ylabel('Absolute error')
    ax2.set_title('Absolute Error Between Beta and Gaussian')
    ax2.grid(alpha=0.3)
    ax2.legend()

    plt.tight_layout()
    plt.show()
```

```python
plot_gaussian_beta([5, 10, 20, 50, 100, 200])
```

```python
plot_gaussian_beta([ 100, 200, 300])
```
